# Aqualia Challenge — ESG materiality
Identified **three emerging material topics** grounded in CSRD/ESRS, scored risk,
operations and strategy impact. Reconstruction with the same scoring grid; the
stakeholder survey layer is simulated (original response files lost).

```python
import sys, os
sys.path.insert(0, os.path.abspath('../src'))
import pandas as pd, matplotlib.pyplot as plt, numpy as np

rng = np.random.default_rng(7)
topics = pd.DataFrame([
    ("E1", "Data-centre water stress in supply basins", "emerging"),
    ("E1", "PFAS & micropollutants in reclaimed water", "emerging"),
    ("E5", "Drought-driven physical asset risk", "emerging"),
    ("S1", "Affordability & water poverty", "classic"),
    ("E3", "Energy intensity of treatment (Scope 1/2)", "classic"),
    ("G1", "AI-assisted leak detection governance", "emerging"),
    ("E4", "Biodiversity of reclaimed-water discharge", "classic"),
], columns=["esrs", "topic", "kind"])
topics["impact"] = rng.uniform(2, 5, len(topics)).round(2)
topics["financial_risk"] = rng.uniform(1.5, 5, len(topics)).round(2)
topics
```

```python
fig, ax = plt.subplots(figsize=(7, 6))
emerging = topics[topics.kind == "emerging"]
classic = topics[topics.kind == "classic"]
ax.scatter(classic.impact, classic.financial_risk, c="#8b94a7", s=90, label="current")
ax.scatter(emerging.impact, emerging.financial_risk, c="#E60012", s=120, label="emerging")
for _, r in topics.iterrows():
    ax.annotate(f'{r.esrs} · {r.topic[:34]}', (r.impact, r.financial_risk), fontsize=7, alpha=.85,
                xytext=(3, 3), textcoords="offset points")
ax.set(xlabel="impact materiality (0–5)", ylabel="financial risk (0–5)", title="Double materiality matrix — Aqualia")
ax.axvline(3, ls="--", c="k", alpha=.3); ax.axhline(3, ls="--", c="k", alpha=.3)
ax.legend()
plt.tight_layout(); plt.show()
```

## Outcome
The three **emerging** topics that crossed the high-impact × rising-risk quadrant and made
the shortlist: data-centre water stress, PFAS in reclaimed water, and AI leak-detection
governance — each mapped to its ESRS disclosure obligation and an operational/strategic
impact note for the disclosure roadmap.
